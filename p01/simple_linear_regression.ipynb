```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('D:\Projects\learning_ml\p01\dataset\student_scores.csv')
```

```python
df.columns
```

```python
# Independent variable (X)
X = df["Hours"].values.reshape(-1, 1)

# Dependent variable (y)
y = df["Scores"].values
```

```python
plt.scatter(X, y)

plt.xlabel("Hours Studied")
plt.ylabel("Scores")
plt.title("Hours vs Scores")

plt.show()
```

```python
model = LinearRegression()
model.fit(X, y)

print("Slope (m):", model.coef_[0])
print("Intercept (b):", model.intercept_)
```

```python
y_pred = model.predict(X)

plt.scatter(X, y)
plt.plot(X, y_pred)

plt.xlabel("Hours")
plt.ylabel("Scores")
plt.title("Simple Linear Regression")

plt.show()
```

```python
# New input
hours = np.array([[9.5]])  # Always 2D array

# Predict
predicted_score = model.predict(hours)

print(f"Predicted score for {hours[0][0]} hours is {predicted_score[0]:.2f}")
```
